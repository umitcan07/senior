# 04 — Audio Quality

Run SNR, clipping, and empty-segment analysis on all WAV files.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path("..").resolve()))

from config import DATA_DIR
from powsm.audio_quality import analyze_audio_file, visualize_segments
import librosa
```

```python
for sentence_dir in sorted(DATA_DIR.iterdir()):
    if not sentence_dir.is_dir() or not sentence_dir.name.isdigit():
        continue

    for wav_file in sorted(sentence_dir.glob("*.wav")):
        print(f"{'=' * 60}")
        print(f"File: {wav_file.relative_to(DATA_DIR)}")
        print(f"{'=' * 60}")

        metrics = analyze_audio_file(str(wav_file))

        print(f"SNR:            {metrics['snr_db']:.2f} dB")
        print(f"Peak:           {metrics['peak_amplitude']:.4f} ({metrics['peak_db']:.2f} dBFS)")
        print(f"Clipping:       {metrics['clipping_ratio']:.4f}% ({metrics['num_clipped']} samples)")
        print(f"Noise segments: {metrics['noise_pct']:.1f}%")
        print(f"Empty segments: {metrics['empty_pct']:.1f}%")
        print()
```
